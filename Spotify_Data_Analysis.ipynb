```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = df.drop(columns=['Unnamed: 0'])

print("Column removed successfully!")
print(df.columns.tolist())
```

```python
df['duration_min'] = df['duration_ms'] / 60000

print(df[['track_name', 'duration_ms', 'duration_min']].head())
```

```python
df.isnull().sum()
```

```python
numeric_cols = df.select_dtypes(include='number').columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].mean()
)

print("Mean imputation completed!")
print(df.isnull().sum())
```

```python
text_cols = ['artists', 'album_name', 'track_name']

df[text_cols] = df[text_cols].fillna('Unknown')

print("Missing text values handled!")
print(df.isnull().sum())
```

```python
top10 = df.sort_values(
    by='popularity',
    ascending=False
)

print(top10[['track_name', 'artists', 'popularity']].head(10))
```

```python
average_popularity = df['popularity'].mean()

print("Average Song Popularity:", round(average_popularity, 2))
```

```python
print("Average Danceability:", round(df['danceability'].mean(), 2))
print("Average Energy:", round(df['energy'].mean(), 2))
```

```python
top10 = df.nlargest(10, 'popularity')

plt.figure(figsize=(12, 6))

plt.bar(top10['track_name'], top10['popularity'])

plt.title('Top 10 Popular Songs on Spotify')
plt.xlabel('Song Name')
plt.ylabel('Popularity Score')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['danceability'], bins=20)

plt.title('Distribution of Song Danceability')
plt.xlabel('Danceability Score')
plt.ylabel('Number of Songs')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['energy'], df['danceability'], alpha=0.3)

plt.title('Energy vs Danceability of Spotify Songs')
plt.xlabel('Energy')
plt.ylabel('Danceability')

plt.show()
```

```python
genre_popularity = df.groupby('track_genre')['popularity'].mean()

top_genres = genre_popularity.sort_values(
    ascending=False
).head(10)

print(top_genres)
```

```python
plt.figure(figsize=(10, 6))

top_genres.plot(kind='bar')

plt.title('Top 10 Genres by Average Popularity')
plt.xlabel('Music Genre')
plt.ylabel('Average Popularity')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
